```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

df = pd.read_csv('mock_coffeeshop.csv', header='infer')

print(df.shape)
df.head(5)
```

```python
#4.1 Opening frequency from 2000 to 2017
opening_frequency = df['Opening Year'].value_counts().sort_index()

opening_frequency
```

```python
years = range(2000, 2018)

opening_frequency = (
    df['Opening Year']
    .value_counts()
    .reindex(years, fill_value=0)
    .sort_index()
)

opening_frequency
```

```python
#Draw the line graph:

opening_frequency.plot(
    kind='line',
    marker='o',
    title='Coffee-Shop Openings by Year'
)

plt.xlabel('Year')
plt.ylabel('Number of Openings')
plt.xticks(list(years))
plt.show()
```

```python
#4.2 Open vs Closed line graphs

#Create separate counts:

open_frequency = (
    df[df['Status'] == 'Open']['Opening Year']
    .value_counts()
    .reindex(years, fill_value=0)
    .sort_index()
)

closed_frequency = (
    df[df['Status'] == 'Closed']['Opening Year']
    .value_counts()
    .reindex(years, fill_value=0)
    .sort_index()
)
```

```python
#Create the figure:

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    years,
    open_frequency,
    marker='o',
    label='Open'
)

ax.plot(
    years,
    closed_frequency,
    marker='o',
    label='Closed'
)

ax.set_title('Coffee-Shop Openings by Status')
ax.set_xlabel('Year')
ax.set_ylabel('Number of Openings')
ax.set_xticks(list(years))
ax.legend()
plt.show()
```

4.3 Description
The graphs show that coffee-shop openings were relatively low in the earlier years and began to increase more sharply during the period where the line rises rapidly. In the last five years, the number of new openings should be compared between the Open and Closed categories. The graphs indicate whether most of the recently opened shops remained open or subsequently closed.
