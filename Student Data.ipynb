```python
import pandas as pd

df = pd.read_csv('mock_studentlist.csv', header='infer')
df.head(5)
```

```python
#1.1 Average height of female learners
df[df['Gender'] == 'F']['Height'].mean()
```

```python
#1.2 Average height of male learners
df[df['Gender'] == 'M']['Height'].mean()
```

```python
#1.3 Average weight of female learners
df[df['Gender'] == 'F']['Weight'].mean()
```

```python
#1.4 Average weight of male learners
df[df['Gender'] == 'M']['Weight'].mean()
```

```python
#1.5 Tallest female learner
df[df['Gender'] == 'F']['Height'].max()
```

```python
#1.6 Shortest male learner
df[df['Gender'] == 'M']['Height'].min()
```

```python
#1.7 Lightest male learner taller than 178 cm
df[(df['Gender'] == 'M') & (df['Height'] > 178)]['Weight'].min()
```

```python
#1.8 Heaviest female learner shorter than 163 cm
df[(df['Gender'] == 'F') & (df['Height'] < 163)]['Weight'].max()
```

```python
#1.9 Average grade of absent learners (Y)
df[df['Absence'] == 'Y']['Grade'].mean()
```

```python
#1.10 Average grade of learners not absent (N)
df[df['Absence'] == 'N']['Grade'].mean()
```

```python
#1.11 Average height of learners with blood type B or O
df[df['Bloodtype'].isin(['B', 'O'])]['Height'].mean()
```

```python
#1.12 Average height of female learners with blood type B or O
df[(df['Gender'] == 'F') & (df['Bloodtype'].isin(['B', 'O']))]['Height'].mean()
```

```python
#1.13 Average age of learners not absent whose grade is at least 2
df[(df['Absence'] == 'N') & (df['Grade'] >= 2)]['Age'].mean()
```
