```python
import pandas as pd

df = pd.read_csv('mock_sales.csv', header='infer')

print(df.shape)
df.head(5)
```

```python
#3.1 Create Revenue
df['Revenue'] = df['Price'] * df['Quantity']

df.head(5)
```

```python
#3.2 Average Price for each Area using groupby()
df.groupby('Area')['Price'].mean()
```

```python
#3.3 Average Price for each Area using pivot_table()
df.pivot_table(values='Price', index='Area', aggfunc='mean')
```

```python
#3.4 Average Price and Quantity for each Area using groupby()
df.groupby('Area')[['Price', 'Quantity']].mean()
```

```python
#3.5 Average Price and Quantity for each Area using pivot_table()
df.pivot_table(
    values=['Price', 'Quantity'],
    index='Area',
    aggfunc='mean'
)
```

```python
#3.6 Total Quantity for each Area and Product
df.pivot_table(
    values='Quantity',
    index='Area',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)
```

```python
#3.7 Total Revenue for each Area and Product
df['Revenue'] = df['Price'] * df['Quantity']

df.pivot_table(
    values='Revenue',
    index='Area',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)
```
