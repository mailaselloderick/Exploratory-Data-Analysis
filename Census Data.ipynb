```python
import pandas as pd
import numpy as np

df = pd.read_csv('mock_census.csv', header='infer')

print(df.shape)
df.head()
```

```python
#2.1 Total population where Province Code is 117 or 118
df['Population'] = df['Population'].str.replace(',', '').astype(int)
df[df['Province Code'].isin([117, 118])]['Population'].sum()
```

```python
#2.2 Average population where Gender Ratio is less than 1
df[df['Gender Ratio'] < 1]['Population'].mean().round(0)
```

```python
#2.3 Places where Gender Ratio < 1 and Person's Household < 2.2
df[(df['Gender Ratio'] < 1) & (df["Person's Household"] < 2.2)]
```

```python
#2.4 Sort from most Households to fewest and display first 10
df.sort_values('Households', ascending=False).head(10)
```
